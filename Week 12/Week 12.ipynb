```python
# IMPORT STATEMENTS
# NONE


# FUNCTION DEFINITIONS
def calculate_rectangle(length, width):
    # calculate the area using length * width
    area = length * width

    # calculate the perimeter using 2 * (length + width)
    perimeter = 2 * (length + width)

    # pack the area and perimeter into a tuple like (area, perimeter)
    result = (area, perimeter)

    # return the tuple
    return result


# PROGRAM LOGIC

# ask the user for the length
length = input("Enter length: ")

# ask the user for the width
width = input("Enter width: ")

# convert both inputs to floats
length = float(length)
width = float(width)

# call the calculate_rectangle function
result = calculate_rectangle(length, width)

# unpack the returned tuple into my_area and my_perimeter
my_area, my_perimeter = result

# print the results clearly for the user
print("\n--- Results ---")
print("The Area is:", my_area)
print("The Perimeter is:", my_perimeter)
```
