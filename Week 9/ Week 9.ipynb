```python
# NAME: rentalcar.py


# IMPORT STATEMENTS
# none needed for this program


# FUNCTION DEFINITIONS

# function get_classification_name(code)
# takes B, S, or L and returns the full name of the car type
def get_classification_name(code):
    if code == "B":
        return "Budget"
    elif code == "S":
        return "Standard"
    elif code == "L":
        return "Luxury"


# function get_daily_rate(code)
# returns the daily price depending on the car type
def get_daily_rate(code):
    if code == "B":
        return 40
    elif code == "S":
        return 60
    elif code == "L":
        return 80


# function calculate_charged_miles(total_miles)
# first 100 miles are free
# return miles over 100
def calculate_charged_miles(total_miles):
    if total_miles > 100:
        return total_miles - 100
    else:
        return 0


# function calculate_bill(code, days, total_miles)
# figures out the total bill based on rate and mileage
def calculate_bill(code, days, total_miles):
    rate = get_daily_rate(code)
    base_cost = rate * days

    charged_miles = calculate_charged_miles(total_miles)
    mileage_cost = charged_miles * 0.35

    total = base_cost + mileage_cost

    return total, charged_miles


# PROGRAM LOGIC

# ask the user for the classification code
# make sure the code is B, S, or L
while True:
    code = input("Enter classification code (B, S, L): ").upper()
    if code in ["B", "S", "L"]:
        break
    else:
        print("Please enter B, S, or L.")


# ask for the starting odometer
start = int(input("Start Odometer: "))

# ask for the ending odometer
end = int(input("End Odometer: "))

# ask how many days the car was rented
days = int(input("Days Rented: "))


# calculate total miles driven
total_miles = end - start
```
