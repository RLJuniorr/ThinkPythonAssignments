```python
# NAME: rentalcar.py


# IMPORT STATEMENTS
# (none needed)


# FUNCTION DEFINITIONS

# function get_classification_name(code)
# take B, S, or L
# return the full name (Budget, Standard, Luxury)
def get_classification_name(code):
    if code == "B":
        return "Budget"
    elif code == "S":
        return "Standard"
    elif code == "L":
        return "Luxury"


# function get_daily_rate(code)
# return price per day based on classification
def get_daily_rate(code):
    if code == "B":
        return 40
    elif code == "S":
        return 60
    elif code == "L":
        return 80


# function calculate_charged_miles(total_miles)
# first 100 miles are free
# return miles over 100
def calculate_charged_miles(total_miles):
    if total_miles > 100:
        return total_miles - 100
    else:
        return 0


# function calculate_bill(code, days, total_miles)
# get daily rate
# calculate base rental cost
# calculate charged miles
# calculate mileage cost
# return total bill and charged miles
def calculate_bill(code, days, total_miles):
    rate = get_daily_rate(code)
    base_cost = rate * days

    charged_miles = calculate_charged_miles(total_miles)
    mileage_cost = charged_miles * 0.35

    total_bill = base_cost + mileage_cost

    return total_bill, charged_miles



# PROGRAM LOGIC

# ask user for classification code
# validate input (must be B, S, L)
while True:
    code = input("Enter classification code (B, S, L): ").upper()
    if code in ["B", "S", "L"]:
        break
    else:
        print("Invalid code. Please enter B, S, or L.")


# ask for start odometer
start_odometer = int(input("Start Odometer: "))


# ask for end odometer
end_odometer = int(input("End Odometer: "))


# ask for days rented
days = int(input("Days Rented: "))


# calculate total miles
total_miles = end_odometer - start_odometer
```
