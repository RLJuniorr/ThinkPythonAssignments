```python
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
def is_palindrome(word):
  word = word.lower()
  reverse = word[::-1]
  if word == reverse:
    return True
  else:
    return False
# PROGRAM LOGIC
word = input("Enter a word: ")

if is_palindrome(word):
  print("That is a palindrome")
else:
  print("That is not a palindrome")
```
