```python
# IMPORT STATEMENTS
# NONE

# FUNCTION DEFINITIONS

def add_players(roster):
    # Ask the user to enter player names
    # Use a while loop so the user can keep adding players
    while True:

        # If the user types "quit", stop the loop
        name = input("Enter player name (or 'quit' to stop): ")
        if name.lower() == "quit":
            break

        # Otherwise add the name to the roster list using append()
        roster.append(name)


def is_on_roster(roster, name):
    # Look through the roster list
    for player in roster:

        # Compare each player to the name being searched
        if player == name:

            # If the player is found, return True
            return True

    # If the loop finishes and the player is not found, return False
    return False


# PROGRAM LOGIC

# Create an empty list called roster
roster = []

# Call add_players to fill the list
add_players(roster)

# Ask the user for one specific name to search for
search_name = input("Who would you like to search for? ")

# Call is_on_roster and store the returned value
result = is_on_roster(roster, search_name)

# Print a message based on the returned value
if result:
    print("Result: Player found!")
else:
    print("Result: Player not found.")
```
