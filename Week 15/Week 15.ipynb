```python
import json
import os

FILE_NAME = "sessions.json"

def load_sessions():
    if os.path.exists(FILE_NAME):
        with open(FILE_NAME, "r") as file:
            return json.load(file)
    return []

def save_sessions(sessions):
    with open(FILE_NAME, "w") as file:
        json.dump(sessions, file)

sessions = load_sessions()

def add_session():
    subject = input("Enter subject: ")
    duration = int(input("Enter study time (minutes): "))

    session = {
        "subject": subject,
        "duration": duration
    }

    sessions.append(session)
    save_sessions(sessions)

    print("Session saved!\n")

def view_sessions():
    if not sessions:
        print("No study sessions yet.\n")
        return

    print("\nStudy Sessions:")
    for i, s in enumerate(sessions):
        print(f"{i+1}. {s['subject']} - {s['duration']} minutes")
    print()

def show_stats():
    if not sessions:
        print("No data to analyze.\n")
        return

    total = sum(s["duration"] for s in sessions)

    subject_totals = {}
    for s in sessions:
        subject_totals[s["subject"]] = subject_totals.get(s["subject"], 0) + s["duration"]

    print(f"\nTotal Study Time: {total} minutes")
    print("Time per Subject:")

    for subject, time in subject_totals.items():
        print(f"- {subject}: {time} minutes")

    print()

def main():
    while True:
        print("1. Add Study Session")
        print("2. View Sessions")
        print("3. View Stats")
        print("4. Exit")

        choice = input("Choose: ")

        if choice == "1":
            add_session()
        elif choice == "2":
            view_sessions()
        elif choice == "3":
            show_stats()
        elif choice == "4":
            print("Goodbye!")
            break
        else:
            print("Invalid choice\n")

if __name__ == "__main__":
    main()
```
