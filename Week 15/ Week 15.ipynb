```python
tasks = []

def add_task():
    name = input("Enter task name: ")
    priority = input("Enter priority (1 = High, 2 = Medium, 3 = Low): ")

    task = {
        "name": name,
        "priority": int(priority)
    }

    tasks.append(task)
    print("Task added!\n")

def view_tasks():
    if not tasks:
        print("No tasks available.\n")
        return

    print("\nYour Tasks:")
    for i, task in enumerate(tasks):
        print(f"{i+1}. {task['name']} (Priority {task['priority']})")
    print()

def complete_task():
    view_tasks()

    if not tasks:
        return

    choice = int(input("Enter task number to complete: ")) - 1

    if 0 <= choice < len(tasks):
        completed = tasks.pop(choice)

        with open("history.txt", "a") as file:
            file.write(completed["name"] + "\n")

        print("Task completed and saved!\n")
    else:
        print("Invalid selection.\n")

def what_is_next():
    for task in tasks:
        if task["priority"] == 1:
            print(f"\nYou should do this next: {task['name']}\n")
            return

    print("\nNo high-priority tasks! You're doing great, keep going!\n")

def main():
    while True:
        print("1. Add Task")
        print("2. View Tasks")
        print("3. Complete Task")
        print("4. What is next?")
        print("5. Exit")

        choice = input("Choose an option: ")

        if choice == "1":
            add_task()
        elif choice == "2":
            view_tasks()
        elif choice == "3":
            complete_task()
        elif choice == "4":
            what_is_next()
        elif choice == "5":
            print("Goodbye!")
            break
        else:
            print("Invalid choice\n")

if __name__ == "__main__":
    main()
```
