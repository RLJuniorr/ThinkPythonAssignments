```python
sessions = []

def add_session():
    subject = input("Enter subject: ")
    duration = int(input("Enter study time (minutes): "))

    session = {
        "subject": subject,
        "duration": duration
    }

    sessions.append(session)
    print("Session added!\n")

def view_sessions():
    if not sessions:
        print("No study sessions yet.\n")
        return

    print("\nStudy Sessions:")
    for i, s in enumerate(sessions):
        print(f"{i+1}. {s['subject']} - {s['duration']} minutes")
    print()

def main():
    while True:
        print("1. Add Study Session")
        print("2. View Sessions")
        print("3. Exit")

        choice = input("Choose: ")

        if choice == "1":
            add_session()
        elif choice == "2":
            view_sessions()
        elif choice == "3":
            print("Goodbye!")
            break
        else:
            print("Invalid choice\n")

if __name__ == "__main__":
    main()
```
