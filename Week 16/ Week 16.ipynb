```python
expenses = []

while True:
    print("\n--- Expense Tracker ---")
    print("1. Add New Expense")
    print("2. Get Totals")
    print("3. Exit")

    choice = input("Selection: ")

    if choice == "1":
        category = input("Enter Category: ")
        vendor = input("Enter Vendor: ")

        try:
            amount = float(input("Enter Amount: "))
        except:
            print("Invalid amount. Try again.")
            continue

        date = input("Enter Date (YYYY-MM-DD): ")

        expense = {
            "category": category,
            "vendor": vendor,
            "amount": amount,
            "date": date
        }

        expenses.append(expense)
        print("Expense added!")

    elif choice == "2":
        print("\n--- Totaling Options ---")
        print("A: All Expenses")
        print("C: Filter by Category")
        print("V: Filter by Vendor")

        option = input("Choice: ").upper()

        total = 0

        if option == "A":
            for e in expenses:
                total += e["amount"]
            print(f"Total of all expenses: ${total:.2f}")

        elif option == "C":
            cat = input("Enter Category to sum: ")
            for e in expenses:
                if e["category"].lower() == cat.lower():
                    total += e["amount"]
            print(f"Total for {cat}: ${total:.2f}")

        elif option == "V":
            ven = input("Enter Vendor to sum: ")
            for e in expenses:
                if e["vendor"].lower() == ven.lower():
                    total += e["amount"]
            print(f"Total for {ven}: ${total:.2f}")

        else:
            print("Invalid option.")

    elif choice == "3":
        print("Goodbye!")
        break

    else:
        print("Invalid selection.")
```
