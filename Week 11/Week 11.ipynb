```python
# IMPORT STATEMENTS
# NONE

# FUNCTION DEFINITIONS

# Function: add_students
# Parameter: gradebook dictionary
# Use a while loop to repeatedly ask for a student name
# If the user types "done", exit the loop
# Otherwise ask for the student's grade (0–100)
# Store the student name as the key and the grade as the value in the dictionary
def add_students(gradebook):

    while True:
        name = input("Enter student name (or 'done'): ")

        if name.lower() == "done":
            break

        grade = int(input(f"Enter {name}'s grade: "))

        gradebook[name] = grade


# Function: calculate_average
# Parameter: gradebook dictionary
# Get all grade values from the dictionary
# Add them together
# Divide by the number of students
# Return the average grade
def calculate_average(gradebook):

    # If no students were entered, return 0 to avoid division by zero
    if len(gradebook) == 0:
        return 0

    total = 0

    for grade in gradebook.values():
        total += grade

    average = total / len(gradebook)

    return average


# PROGRAM LOGIC

# Create an empty dictionary called gradebook
gradebook = {}

# Call add_students to fill the dictionary
add_students(gradebook)

# Call calculate_average to compute the class average
average = calculate_average(gradebook)

# Print the final report
print("\n--- Class Report ---")

for student, grade in gradebook.items():
    print(f"{student}: {grade}")

# Print the class average formatted to two decimal places
print(f"Class Average: {average:.2f}")
```
